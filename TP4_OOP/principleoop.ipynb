```python
#Abstraction
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

print(Square(4).area())    # 16
```

```python

#Inheritance
class Animal:
    def eat(self):
        print("This animal eats food.")

class Dog(Animal):
    pass

Dog().eat()    # This animal eats food.
```

```python
#Polymorphism
class Cat:
    def speak(self):
        print("Meow")

class Dog:
    def speak(self):
        print("Woof")

for animal in (Cat(), Dog()):
    animal.speak()    # Meow, then Woof
```
