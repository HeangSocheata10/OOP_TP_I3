```python
# Create a class Book with title, author, and price attributes.
# Demonstrate encapsulation by protecting the price attribute.
# Use inheritance to create an EBook subclass.

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = None
        self.set_price(price) 

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            raise ValueError("Price should be a non-negative value.")

    def display(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Price:", self.get_price())


class EBook(Book):
    def __init__(self, title, author, price, file_format):
        super().__init__(title, author, price)
        self.file_format = file_format

    def display(self):
        super().display()
        print("File Format:", self.file_format)


# Get Book input
print("Enter Book Information")
title = input("Enter book title: ")
author = input("Enter author name: ")
price = float(input("Enter book price: "))

book1 = Book(title, author, price)

print("\n---Book Information---")
book1.display()

# Get EBook input
print("\nEnter EBook Information")
ebook_title = input("Enter ebook title: ")
ebook_author = input("Enter author name: ")
ebook_price = float(input("Enter ebook price: "))
ebook_file_format = input("Enter ebook file format: ")

book2 = EBook(ebook_title, ebook_author, ebook_price, ebook_file_format)

print("\n---EBook Information---")
book2.display()
```
